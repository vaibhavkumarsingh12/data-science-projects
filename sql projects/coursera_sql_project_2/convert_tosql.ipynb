```python
import pandas as pd
import sqlite3

# Load your Excel file (change filename as needed)
excel_file = "dataset.xlsx"

# Check all sheet names
xl = pd.ExcelFile(excel_file)
print("Sheets found:", xl.sheet_names)

# Connect to SQLite (creates the .db file automatically)
conn = sqlite3.connect("project.db")

# Load each sheet as a separate SQL table
for sheet in xl.sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet)
    table_name = sheet.replace(" ", "_").lower()  # clean table names
    df.to_sql(table_name, conn, if_exists="replace", index=False)
    print(f"✅ Table '{table_name}' created with {len(df)} rows")

conn.close()
print("\nDone! Your database is ready: project.db")
```

```python
# Conditional Tests - Exercise 5-1

car = 'subaru'
name = 'alice'
age = 25
fruits = ['apple', 'banana', 'mango']
number = 42

# ── Test 1 ──────────────────────────────────────────────
print("Is car == 'subaru'? I predict True.")
```

```python
print(car == 'subaru')

# ── Test 2 ──────────────────────────────────────────────
print("\nIs car == 'honda'? I predict False.")
print(car == 'honda')

# ── Test 3 ──────────────────────────────────────────────
print("\nIs name == 'alice'? I predict True.")
print(name == 'alice')

# ── Test 4 ──────────────────────────────────────────────
print("\nIs name.upper() == 'ALICE'? I predict True.")
print(name.upper() == 'ALICE')

# ── Test 5 ──────────────────────────────────────────────
print("\nIs age == 30? I predict False.")
print(age == 30)

# ── Test 6 ──────────────────────────────────────────────
print("\nIs age >= 18? I predict True.")
print(age >= 18)

# ── Test 7 ──────────────────────────────────────────────
print("\nIs age != 25? I predict False.")
print(age != 25)

# ── Test 8 ──────────────────────────────────────────────
print("\nIs number > 100? I predict False.")
print(number > 100)

# ── Test 9 ──────────────────────────────────────────────
print("\nIs 'banana' in fruits? I predict True.")
print('banana' in fruits)

# ── Test 10 ─────────────────────────────────────────────
print("\nIs 'grape' in fruits? I predict False.")
print('grape' in fruits)

# ── Test 11 ─────────────────────────────────────────────
print("\nIs (age > 18 and number == 42)? I predict True.")
print(age > 18 and number == 42)

# ── Test 12 ─────────────────────────────────────────────
print("\nIs (car == 'honda' or car == 'subaru')? I predict True.")
print(car == 'honda' or car == 'subaru')
```
